```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df['Month'] = df['Order_Date'].dt.to_period('M')
df[['Order_Date','Month']].head()
```

```python
total_sales = df['Total_Sales'].sum()
print("Total Sales =",total_sales)
```

```python
total_profit = df['Profit'].sum()

print("Total Profit =",total_profit)
```

```python
total_orders = df['Order_ID'].unique()
print("Total Orders =",total_orders)
```

```python
total_customers = df['Customer_Name'].unique()
print("Total Customers =",total_customers)
```

```python
top_products = (df.groupby('Product_Name')['Total_Sales']
                .sum().sort_values(ascending=False).head(10)
)

top_products
```

```python
#Visualize Top Products

top_products.plot(kind= 'bar',figsize=(10,8))

plt.title('Top 10 Products by Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
category_sales = (df.groupby('Product_Category')['Total_Sales']
                 .sum()
                 .sort_values(ascending=False)
)

category_sales
```

```python
#Visualize Category Sales

category_sales.plot(kind='bar',figsize=(8,5))

plt.title('Category Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
category_profit = (
    df.groupby('Product_Category')['Profit']
      .sum()
      .sort_values(ascending=False)
)

category_profit
```

```python
#Visualize Category Profit

category_profit.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Profit by Product Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

```python
region_sales = (
    df.groupby('Region')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
)

region_sales
```

```python
#Visualize Regional Sales

region_sales.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
country_sales = (
    df.groupby('Country')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

country_sales
```

```python
#Visualize Top Countries

country_sales.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Top 10 Countries by Sales')
plt.xlabel('Country')
plt.ylabel('Sales')
plt.show()

```

```python
monthly_sales = (
    df.groupby('Month')['Total_Sales']
      .sum()
)

monthly_sales
```

```python
#Visualize Monthly Trend

monthly_sales.plot(
    figsize=(12,5)
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
#Top 10 Customers

top_customers = (
    df.groupby('Customer_Name')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

```python
#Visualize Top Customers

top_customers.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Top 10 Customers by Sales')
plt.xlabel('Customer')
plt.ylabel('Sales')
plt.show()
```

```python
#Customer Segment Analysis
segment_sales = (
    df.groupby('Customer_Segment')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
)

segment_sales
```

```python
#Visualize Customer Segments

segment_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.ylabel('')
plt.title('Sales by Customer Segment')
plt.show()
```

```python
#Payment Method Analysis
payment_sales = (
    df.groupby('Payment_Method')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
)

payment_sales

```

```python
#Visualize Payment Methods

payment_sales.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Sales')
plt.show()
```

```python
#Discount Analysis
plt.figure(figsize=(8,5))

plt.scatter(
    df['Discount_Percent'],
    df['Profit']
)

plt.title('Discount vs Profit')
plt.xlabel('Discount Percent')
plt.ylabel('Profit')

plt.show()
```

```python
#Correlation Analysis

df[['Quantity',
    'Unit_Price',
    'Discount_Percent',
    'Total_Sales',
    'Shipping_Cost',
    'Profit']].corr()
```

```python
#Heatmap

plt.figure(figsize=(8,6))

sns.heatmap(
    df[['Quantity',
        'Unit_Price',
        'Discount_Percent',
        'Total_Sales',
        'Shipping_Cost',
        'Profit']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```
